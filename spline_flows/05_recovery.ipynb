```python
import os, sys, json
import numpy as np
import matplotlib.pyplot as plt

PROJECT = os.path.expanduser("~/spline_flows")
FIG = os.path.join(PROJECT, "figures")
sys.path.insert(0, PROJECT)
import common as C
import recovery as R

d = np.load(C.DATA)
s = np.load(C.SPLINE)
idx = np.random.default_rng(0).choice(len(d["Y_test"]), size=300, replace=False)
refs = {
    "Spline": R.spline_reference(s["W_test"][idx], d["J_min"], d["J_max"]),
    "Waypoint": R.waypoint_reference(d["Y_test"][idx]),
}
P = refs["Spline"][0]
path_len = np.linalg.norm(np.diff(P, axis=1), axis=-1).sum(1)
gap = np.linalg.norm(P[:, -1] - P[:, 0], axis=-1)
print(f"{len(idx)} test windows, path length {np.median(path_len):.2f} rad (median), "
      f"start-to-end gap {np.median(gap):.2f} rad")

COLOR = {"Spline": "#2a78d6", "Waypoint": "#eb6834"}
STYLE = {"tracking": "--", "flow": "-"}
K_GAIN = 6.0
```

```python
for name, (pos, vel) in refs.items():
    for ctrl in ["tracking", "flow"]:
        _, dist, _ = R.simulate(pos, ctrl, np.zeros((len(pos), C.J)), k=K_GAIN)
        print(f"no push, {name}, {ctrl}: max distance {dist.max():.1e} rad")
```

```python
MAGS = [0.1, 0.2, 0.4, 0.8, 1.6, 3.2]
KINDS = ["random", "perpendicular", "forward", "backward", "one joint"]
results = []
for kind in KINDS:
    for m in MAGS:
        for name, (pos, vel) in refs.items():
            push = R.push_vectors(pos, vel, kind, m)
            for ctrl in ["tracking", "flow"]:
                xs, dist, phase = R.simulate(pos, ctrl, push, k=K_GAIN)
                met = R.metrics(xs, dist, phase, pos, m)
                results.append({
                    "kind": kind, "magnitude": m, "rep": name, "controller": ctrl,
                    "recovery_time_s": float(np.nanmedian(met["recovery_time_s"])),
                    "not_recovered": float(np.mean(np.isnan(met["recovery_time_s"]))),
                    "phase_jump_s": float(np.median(met["phase_jump_s"])),
                    "big_jump_frac": float(np.mean(np.abs(met["phase_jump_s"]) > 0.1)),
                    "phase_shift_s": float(np.median(met["phase_shift_s"])),
                    "clock_error_rad": float(np.median(met["clock_error_rad"])),
                    "rms_jerk": float(np.median(met["rms_jerk"])),
                })
with open(os.path.join(PROJECT, "runs", "recovery_results.json"), "w") as f:
    json.dump(results, f, indent=1)
print(len(results), "settings")


def get(kind, m, rep, ctrl, key):
    return next(r[key] for r in results if r["kind"] == kind and r["magnitude"] == m
                and r["rep"] == rep and r["controller"] == ctrl)
```

```python
M = 0.4
print(f"push size {M} rad")
print(f"{'':22s}{'recovery (s)':>14s}{'phase shift fwd/back (s)':>27s}{'clock error back (rad)':>25s}{'RMS jerk':>10s}")
for rep in refs:
    for ctrl in ["tracking", "flow"]:
        print(f"{rep + ', ' + ctrl:22s}"
              f"{get('perpendicular', M, rep, ctrl, 'recovery_time_s'):14.3f}"
              f"{get('forward', M, rep, ctrl, 'phase_shift_s'):+15.3f} / {get('backward', M, rep, ctrl, 'phase_shift_s'):+.3f}"
              f"{get('backward', M, rep, ctrl, 'clock_error_rad'):23.3f}"
              f"{get('perpendicular', M, rep, ctrl, 'rms_jerk'):12.0f}")
```

```python
w = 0
t = np.arange(int(round((R.T_REF + 0.5) / R.DT)) + 1) * R.DT
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for name, (pos, vel) in refs.items():
    push = R.push_vectors(pos, vel, "perpendicular", 0.4)
    for ctrl in ["tracking", "flow"]:
        _, dist, _ = R.simulate(pos[w:w + 1], ctrl, push[w:w + 1], k=K_GAIN)
        axes[0].semilogy(t, dist[0] + 1e-6, color=COLOR[name], ls=STYLE[ctrl], lw=2, label=f"{name}, {ctrl}")
tp = t[t >= 0.2]
axes[0].semilogy(tp, 0.4 * (1 - K_GAIN * R.DT) ** np.arange(len(tp)), color="black", lw=1, ls=":",
                 label="0.4 (1 - k dt)^n")
axes[0].set(title="Distance to the curve after a 0.4 rad sideways push", xlabel="time (s)",
            ylabel="distance (rad, log scale)", ylim=(1e-4, 1))
axes[0].legend(fontsize=8, frameon=False)

pos, vel = refs["Spline"]
push = R.push_vectors(pos, vel, "backward", 0.8)
for ctrl in ["tracking", "flow"]:
    _, _, phase = R.simulate(pos[w:w + 1], ctrl, push[w:w + 1], k=K_GAIN)
    axes[1].plot(t, phase[0], color=COLOR["Spline"], ls=STYLE[ctrl], lw=2, label=f"Spline, {ctrl}")
axes[1].plot(t, np.minimum(t, R.T_REF), color="#9a9893", lw=1, ls=":", label="clock")
axes[1].set(title="Progress after a 0.8 rad backward push", xlabel="time (s)",
            ylabel="closest point on the curve (s along it)")
axes[1].legend(fontsize=8, frameon=False)
for ax in axes:
    ax.grid(alpha=0.3)
    ax.axvline(0.2, color="#c3c2b7", lw=1)
fig.tight_layout()
fig.savefig(os.path.join(FIG, "recovery_example.png"), dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for rep in refs:
    for ctrl in ["tracking", "flow"]:
        kw = dict(color=COLOR[rep], ls=STYLE[ctrl], lw=2, marker="o", ms=4, label=f"{rep}, {ctrl}")
        axes[0].plot(MAGS, [get("random", m, rep, ctrl, "recovery_time_s") for m in MAGS], **kw)
        axes[1].plot(MAGS, [get("backward", m, rep, ctrl, "phase_shift_s") for m in MAGS], **kw)
        axes[2].plot(MAGS, [get("random", m, rep, ctrl, "big_jump_frac") for m in MAGS], **kw)
axes[0].set(title="Recovery time, random pushes", ylabel="time to 10% of push (s)", ylim=(0, 0.6))
axes[1].set(title="Phase shift after backward pushes", ylabel="ahead (+) / behind (-) the clock (s)")
axes[2].set(title="Closest point jumps > 0.1 s, random pushes", ylabel="fraction of windows")
for ax in axes:
    ax.set_xscale("log")
    ax.set_xticks(MAGS, [str(m) for m in MAGS])
    ax.set_xlabel("push size (rad)")
    ax.grid(alpha=0.3)
axes[0].legend(fontsize=8, frameon=False)
fig.tight_layout()
fig.savefig(os.path.join(FIG, "recovery_sweep.png"), dpi=150)
plt.show()
```
